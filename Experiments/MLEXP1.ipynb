```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside notebook
%matplotlib inline
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
df.head(10)
```

```python
df.tail(5)
```

```python
print(df.shape)
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Age")
print("Mean :", df["Age"].mean())
print("Minimum :", df["Age"].min())
print("Maximum :", df["Age"].max())

print()

print("Fare")
print("Mean :", df["Fare"].mean())
print("Minimum :", df["Fare"].min())
print("Maximum :", df["Fare"].max())
```

```python
df.isnull().sum()
```

```python
median_age = df["Age"].median()

df["Age"] = df["Age"].fillna(median_age)

df["Age"].isnull().sum()
```

```python
survivors = df[df["Survived"] == 1]

survivors.head()
```

```python
fare_array = df["Fare"].to_numpy()

print(fare_array)
```

```python
print("Maximum Fare :", np.max(fare_array))
print("Average Fare :", np.mean(fare_array))
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Age"], bins=20, color="skyblue", edgecolor="black")

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="Survived", data=df)

plt.title("Survival Count")
plt.xlabel("Survived")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Age"], df["Fare"], alpha=0.6)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

Better Looking Plots with Seaborn

```python
sns.histplot(df["Age"], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.scatterplot(x="Age", y="Fare", data=df)
plt.title("Age vs Fare")
plt.show()
```
